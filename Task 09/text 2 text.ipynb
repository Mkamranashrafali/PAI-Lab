```python
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
model="google/flan-t5-small"
tokenizer = AutoTokenizer.from_pretrained(model)
model = AutoModelForSeq2SeqLM.from_pretrained(model)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)


def generate(usertext,length=60):
    prefix = "summarize  : "
    prompt = prefix +usertext
    # Tokenize the prompt and move to device
    tokenized_input = tokenizer(prompt,return_tensors="pt")
    input_ids = tokenized_input.input_ids.to(device)
    attention_mask = tokenized_input.attention_mask.to(device)

    outputs = model.generate(
        input_ids=input_ids,
        attention_mask=attention_mask,
        max_length = length,
    )
    return tokenizer.decode(outputs[0],skip_special_tokens=True)



generate("Allama Iqbal was a great philosopher, poet, and thinker of the Muslim world. He inspired Muslims to believe in themselves and work for their rights. His poetry gave a message of hope, self-respect, and unity. Iqbal played an important role in the idea of Pakistan. He is known as the “Poet of the East” and is highly respected in history.")

```
